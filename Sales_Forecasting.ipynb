```python
import os
import pandas as pd
import numpy
import seaborn as sns
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/drive')
dt = pd.read_csv('/content/drive/MyDrive/Train.csv')
item_weight_mean = dt.pivot_table(values = "Item_Weight", index = 'Item_Identifier')
display (item_weight_mean)
```

```python
display (dt['Item_Identifier'])
```

```python
import numpy as np

# Fill missing Item_Weight with the mean of the Item_Identifier group.
# For Item_Identifiers where all Item_Weight values are NaN, this will result in NaN.
dt['Item_Weight'] = dt.groupby('Item_Identifier')['Item_Weight'].transform(lambda x: x.fillna(x.mean()))

# Fill any remaining missing Item_Weight (for Item_Identifiers that had all their weights missing)
# with the global mean of Item_Weight.
dt['Item_Weight'] = dt['Item_Weight'].fillna(dt['Item_Weight'].mean())
```

```python
result = dt['Item_Weight'].isnull().sum()
display (result)
```

```python
result = dt.groupby('Outlet_Size').agg({'Outlet_Size': np.size})
display (result)
```

```python
result= dt.groupby (['Outlet_Type','Outlet_Size'] ).agg({'Outlet_Type':[np.size]})
display (result)
```

```python
outlet_size_mode = dt.pivot_table(values='Outlet_Size', columns='Outlet_Type', aggfunc=(lambda x: x.mode()[0]))
display (outlet_size_mode)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
dt['Outlet'] = le.fit_transform(dt['Outlet_Identifier'])
display (dt['Outlet'])
```

```python
cat_col = ['Item_Fat_Content', 'Item_Type', 'Outlet_Size', 'Outlet_Location_Type', 'Outlet_Type', 'New_Item_Type']
for col in cat_col:
    dt[col] = le.fit_transform(dt[col])
display (dt.head())
```

```python
dt = pd.get_dummies(dt, columns=['Item_Fat_Content', 'Outlet_Size', 'Outlet_Location_Type', 'Outlet_Type', 'New_Item_Type'],dtype = int )
display (dt.head())
```

```python
X = dt.drop(columns=['Outlet_Establishment_Year', 'Item_Identifier', 'Outlet_Identifier', 'Item_Outlet_Sales'])
display (X.head())
```

```python
y = dt['Item_Outlet_Sales']
display (y.head())
```

```python
from sklearn import metrics
display (",   ".join(metrics.get_scorer_names()))
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split

def train(model, X, y):
    print ("Train Test Split")
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

    model.fit(X_train, y_train)

    print (X.shape, y.shape)
    print (X_train.shape, X_test.shape ,  y_train.shape, y_test.shape)

    return model, X_train, X_test, y_train, y_test
```

```python
from sklearn.linear_model import LinearRegression
lr_model = LinearRegression()

lr_model, X_train, X_test, y_train, y_test = train(lr_model, X, y)
```

```python
y_pred = lr_model.predict(X_test)
display(y_pred)
```

```python
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error: {mse:.4f}")
print(f"R-squared: {r2:.4f}")
```

```python
cv_score = cross_val_score(lr_model, X, y, scoring='neg_mean_squared_error', cv=5)
print("Model Report")
print ('Scoring - neg_mean_squared_error')
print ( cv_score )
cv_score = np.abs(np.mean(cv_score))
print ('ABS Average of - neg_mean_squared_error',cv_score )
cv_score = cross_val_score(lr_model, X, y,  cv=5)
print ()
print ('R2 Score ')
print ( cv_score )
cv_score = np.mean(cv_score)
print ('Average R2 Score ',cv_score)
print ()
```

```python
print ('Accuracy')
print ('Accuracy of Test data')
y_test_pred = lr_model.predict(X_test)
print('R2_Score:', r2_score(y_test,y_test_pred))
print ('Accuracy of Training data')
y_train_pred = lr_model.predict(X_train)
print('R2_Score:', r2_score(y_train,y_train_pred))
print ('Accuracy of Complete data')
y_pred = lr_model.predict(X)
print('R2_Score:', r2_score(y,y_pred))
print ()
```

```python
plt.subplot (212)
print ('Display actual and predicted values')
sns.regplot( x =y, y= y_pred, scatter_kws={"color": "b"},
            line_kws={"color": "r"},ci = None)
plt.show()
```

```python
random_grid = {
               'max_features': max_features,
               'max_depth': max_depth,
               'min_samples_split': min_samples_split,
               'min_samples_leaf': min_samples_leaf}

print(random_grid)
```
